```python
import numpy as np
data=np.array([[10,20,30],
              [20,30,40],
              [40,50,60]])
print(data)
row_sum=np.sum(data,axis=1)
print("sum of rows:",row_sum)
column_sum = np.sum(data, axis=0)
print("sum of column:", column_sum)
mean = np.mean(data)
print("Mean:", mean)
median = np.median(data)
print("Median:", median)
values, counts = np.unique(data, return_counts=True)
mode = values[np.argmax(counts)]
print("Mode:", mode)
std = np.std(data)
print("Standard Deviation:", std)



```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create values from 0 to 2π
x = np.linspace(0, 2 * np.pi, 100)

# Calculate sine values
y = np.sin(x)

# Plot the sine curve
plt.plot(x, y)

# Add labels and title
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.title("Sine Curve")

# Display the graph
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.linspace(0, 10, 500)

# Damped sine curve
y = np.exp(-0.3 * x) * np.sin(2 * np.pi * x)

# Plot the curve
plt.plot(x, y)

# Add labels
plt.xlabel("X")
plt.ylabel("Amplitude")
plt.title("Damped Sine Curve")

# Show grid
plt.grid()

# Display graph
plt.show()
```
