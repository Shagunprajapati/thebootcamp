```python
import cv2

# 1. Read image
image = cv2.imread(r"C:/Users/SHAGUN PRAJAPATI/car.jpg")

if image is None:
    print("Image not found!")
else:
    print("Image loaded successfully")

    # 2. Resize image
    resized = cv2.resize(image, (500, 400))

    # 3. Convert to grayscale
    gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

    # 4. Crop image
    cropped = resized[50:350, 100:400]

    # 5. Rotate image
    rotated = cv2.rotate(
        resized,
        cv2.ROTATE_90_CLOCKWISE
    )

    # 6. Blur image
    blurred = cv2.GaussianBlur(
        resized,
        (7, 7),
        0
    )

    # 7. Detect edges
    edges = cv2.Canny(
        gray,
        100,
        200
    )

    # 8. Increase brightness
    bright = cv2.convertScaleAbs(
        resized,
        alpha=1.0,
        beta=50
    )

    # 9. Save edited images
    cv2.imwrite("resized.jpg", resized)
    cv2.imwrite("gray.jpg", gray)
    cv2.imwrite("cropped.jpg", cropped)
    cv2.imwrite("rotated.jpg", rotated)
    cv2.imwrite("blurred.jpg", blurred)
    cv2.imwrite("edges.jpg", edges)
    cv2.imwrite("bright.jpg", bright)

    print("All edited images saved successfully!")
```
